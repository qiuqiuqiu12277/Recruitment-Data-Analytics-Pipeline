# Recruitment Data Analytics EDA

This notebook uses the same validated package API as the command-line workflow. The repository data is synthetic and contains no applicant information. Install the project with `python -m pip install -e ".[dev]"` before running the cells.

```python
import json
from pathlib import Path

import pandas as pd

from recruitment_pipeline import run_pipeline

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
```

```python
result = run_pipeline(
    PROJECT_ROOT / "data" / "raw_jobs.csv",
    PROJECT_ROOT / "outputs" / "notebook",
    include_charts=False,
)
result
```

```python
jobs = pd.read_csv(result.processed_csv)
jobs["skills"] = jobs["skills"].map(json.loads)
summary = json.loads(result.summary_json.read_text(encoding="utf-8"))
display(pd.DataFrame([summary["engagement"]]))
display(jobs.head())
```

```python
ax = (
    jobs["skills"]
    .explode()
    .dropna()
    .value_counts()
    .head(10)
    .sort_values()
    .plot.barh(title="Most requested skills", figsize=(8, 4))
)
ax.set_xlabel("Number of job postings");
```

## Interpretation notes

The included sample is intentionally small and demonstrates the workflow rather than a representative labour-market estimate. Use larger, legally obtained and properly anonymized data before drawing real-world conclusions.
